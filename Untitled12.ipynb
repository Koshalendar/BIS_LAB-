```python
import numpy as np

def fitness_function(wolf_position, start, goal, obstacles, num_checkpoints):
    points = wolf_position.reshape(num_checkpoints, 2)
    full_path = np.vstack([start, points, goal])
    distance = sum(np.linalg.norm(full_path[i] - full_path[i+1]) for i in range(len(full_path) - 1))
    penalty = sum(10000 for pt in points for obs in obstacles if np.linalg.norm(pt - np.array(obs[:2])) <= obs[2])
    return distance + penalty

def gwo_path_planner(start, goal, obstacles, num_checkpoints, min_bound, max_bound, num_wolves, max_iter):
    dims = num_checkpoints * 2
    wolves = np.random.uniform(min_bound, max_bound, (num_wolves, dims))
    alpha_pos, alpha_score = np.zeros(dims), float('inf')
    beta_pos, beta_score = np.zeros(dims), float('inf')
    delta_pos, delta_score = np.zeros(dims), float('inf')

    for t in range(max_iter):
        for i in range(num_wolves):
            wolves[i] = np.clip(wolves[i], min_bound, max_bound)
            score = fitness_function(wolves[i], start, goal, obstacles, num_checkpoints)
            if score < alpha_score:
                alpha_score, alpha_pos = score, wolves[i].copy()
            elif score < beta_score:
                beta_score, beta_pos = score, wolves[i].copy()
            elif score < delta_score:
                delta_score, delta_pos = score, wolves[i].copy()

        a = 2 - (t * (2 / max_iter))
        for i in range(num_wolves):
            X1 = alpha_pos - (2 * a * np.random.rand(dims) - a) * np.abs((2 * np.random.rand(dims)) * alpha_pos - wolves[i])
            X2 = beta_pos - (2 * a * np.random.rand(dims) - a) * np.abs((2 * np.random.rand(dims)) * beta_pos - wolves[i])
            X3 = delta_pos - (2 * a * np.random.rand(dims) - a) * np.abs((2 * np.random.rand(dims)) * delta_pos - wolves[i])
            wolves[i] = (X1 + X2 + X3) / 3

    return np.vstack([start, alpha_pos.reshape(num_checkpoints, 2), goal]), alpha_score

if __name__ == "__main__":
    start_x = float(input("Enter Start X: "))
    start_y = float(input("Enter Start Y: "))
    goal_x = float(input("Enter Goal X: "))
    goal_y = float(input("Enter Goal Y: "))

    start = np.array([start_x, start_y])
    goal = np.array([goal_x, goal_y])

    num_obs = int(input("Enter number of obstacles: "))
    obstacles = []
    for i in range(num_obs):
        x = float(input(f"Obstacle {i+1} X: "))
        y = float(input(f"Obstacle {i+1} Y: "))
        r = float(input(f"Obstacle {i+1} Radius: "))
        obstacles.append((x, y, r))

    checkpoints = int(input("Enter number of path checkpoints: "))
    min_b = float(input("Enter minimum map boundary: "))
    max_b = float(input("Enter maximum map boundary: "))
    wolves = int(input("Enter wolf pack size (e.g. 30): "))
    iters = int(input("Enter maximum iterations (e.g. 100): "))

    path, cost = gwo_path_planner(start, goal, obstacles, checkpoints, min_b, max_b, wolves, iters)
    print("\nOptimized Path Coordinates:\n", path)
    print(f"Final Path Score: {cost:.2f}")
```
